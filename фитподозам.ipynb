```python
# -*- coding: utf-8 -*-
"""
C+ model: separate fit for each dose
NO F_FAST

For each dose:
- fit gamma and neutron jointly
- common kinetic/signaling parameters within this dose
- separate initial fast fractions alpha_g / alpha_n
- separate ultraslow fractions pi_g / pi_n

States:
    [D_s, D_c, D_u, pATM_f, pATM_s, gH2AX]
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import least_squares
import pandas as pd
import os
import json
from IPython.display import display

# =========================================
# Settings
# =========================================
SAVE_DIR = "./Cplus_per_dose_clean_alpha"
os.makedirs(SAVE_DIR, exist_ok=True)

plt.rcParams.update({
    'figure.figsize': (12, 5),
    'font.size': 12,
    'axes.labelsize': 13,
    'axes.grid': True,
    'grid.alpha': 0.3,
    'legend.frameon': False
})

# =========================================
# Experimental data
# =========================================
t_exp = np.array([0.5, 2.0, 4.0, 6.0, 24.0], dtype=float)
DOSES = [0.25, 0.5, 1.0, 2.0]

DATA = {

    0.25: {
        "gamma": {
            "gh2ax":     np.array([8.12, 6.41, 4.82, 4.16, 2.21], dtype=float),
            "gh2ax_err": np.array([0.69, 0.39, 0.48, 0.52, 0.43], dtype=float),
            "patm":      np.array([3.26, 2.48, 1.92, 1.56, 0.80], dtype=float),
            "patm_err":  np.array([0.44, 0.44, 0.49, 0.39, 0.26], dtype=float),
        },
        "neutron": {
            "gh2ax":     np.array([7.69, 7.16, 6.17, 5.66, 4.96], dtype=float),
            "gh2ax_err": np.array([0.41, 0.44, 0.51, 0.66, 0.39], dtype=float),
            "patm":      np.array([4.43, 3.87, 3.41, 3.12, 2.01], dtype=float),
            "patm_err":  np.array([0.31, 0.73, 0.61, 0.40, 0.39], dtype=float),
        },
    },

    0.5: {
        "gamma": {
            "gh2ax":     np.array([11.90, 8.85, 7.01, 6.06, 2.65]),
            "gh2ax_err": np.array([0.31, 0.56, 0.51, 0.48, 0.14]),
            "patm":      np.array([6.91, 4.65, 3.56, 2.83, 0.93]),
            "patm_err":  np.array([0.39, 0.56, 0.43, 0.53, 0.25]),
        },
        "neutron": {
            "gh2ax":     np.array([10.63, 9.51, 8.86, 7.57, 6.00]),
            "gh2ax_err": np.array([0.80, 0.37, 0.63, 0.47, 0.53]),
            "patm":      np.array([6.49, 4.72, 3.95, 3.53, 2.41]),
            "patm_err":  np.array([0.33, 0.64, 0.82, 0.51, 0.68]),
        },
    },

    1.0: {
        "gamma": {
            "gh2ax":     np.array([23.07, 13.56, 10.92, 8.23, 2.54]),
            "gh2ax_err": np.array([1.75, 0.67, 1.57, 0.59, 0.25]),
            "patm":      np.array([9.98, 7.05, 4.69, 3.35, 1.21]),
            "patm_err":  np.array([0.94, 0.62, 0.45, 0.61, 0.19]),
        },
        "neutron": {
            "gh2ax":     np.array([19.28, 15.65, 11.11, 8.98, 6.46]),
            "gh2ax_err": np.array([1.63, 0.84, 1.45, 0.97, 1.21]),
            "patm":      np.array([8.78, 7.01, 5.79, 5.16, 2.96]),
            "patm_err":  np.array([0.36, 0.55, 0.60, 0.67, 0.22]),
        },
    },

    2.0: {
        "gamma": {
            "gh2ax":     np.array([33.8, 18.61, 15.84, 12.53, 2.67]),
            "gh2ax_err": np.array([2.13, 2.29, 2.07, 1.68, 0.57]),
            "patm":      np.array([18.32, 12.66, 6.45, 4.62, 1.33]),
            "patm_err":  np.array([2.12, 0.60, 0.60, 0.49, 0.24]),
        },
        "neutron": {
            "gh2ax":     np.array([25.80, 20.45, 17.11, 12.26, 7.97]),
            "gh2ax_err": np.array([2.11, 1.99, 0.90, 1.22, 1.50]),
            "patm":      np.array([14.09, 12.21, 11.14, 8.56, 3.77]),
            "patm_err":  np.array([1.31, 0.79, 0.80, 0.97, 0.49]),
        },
    },
}

# =========================================
# Model C+
# =========================================
def rhs_cplus(t, y, p):
    D_s, D_c, D_u, pATM_f, pATM_s, gH2AX = y

    dD_s = -p['k_rs'] * D_s
    dD_c = -(p['V_rc'] * D_c) / (p['K_rc'] + D_c + 1e-12) - p['k_cu'] * D_c
    dD_u = -p['k_u'] * D_u + p['k_cu'] * D_c

    w_du = p['s_du'] * p['w_dc']
    weighted = D_s + p['w_dc'] * D_c + w_du * D_u

    prod_f = (p['V_af'] * weighted) / (p['K_m'] + weighted + 1e-12)
    dpATM_f = prod_f - p['k_df'] * pATM_f
    dpATM_s = p['k_as'] * pATM_f - p['k_ds'] * pATM_s

    dgH2AX = p['k_p'] * (pATM_f + pATM_s) - p['k_g'] * gH2AX

    return [dD_s, dD_c, dD_u, dpATM_f, dpATM_s, dgH2AX]


def simulate_cplus(dose, rad, t_eval, p):
    alpha_fast = p['alpha_g'] if rad == "gamma" else p['alpha_n']
    alpha_fast = float(np.clip(alpha_fast, 0.02, 0.95))

    pi = p['pi_g'] if rad == "gamma" else p['pi_n']
    pi = float(np.clip(pi, 0.0, 0.95))

    D_s0 = alpha_fast * dose
    D_slow0 = (1.0 - alpha_fast) * dose

    D_u0 = pi * D_slow0
    D_c0 = (1.0 - pi) * D_slow0

    y0 = [D_s0, D_c0, D_u0, 0.0, 0.0, 0.0]

    sol = solve_ivp(
        lambda t, y: rhs_cplus(t, y, p),
        (0.0, float(np.max(t_eval))),
        y0,
        t_eval=t_eval,
        method='LSODA',
        rtol=1e-6,
        atol=1e-8
    )
    return sol

# =========================================
# Parameterization for one dose
# =========================================
PARAM_NAMES = [
    'k_rs', 'V_rc', 'K_rc',
    'V_af', 'K_m', 'k_df', 'k_as', 'k_ds', 'k_p', 'k_g',
    'w_dc', 'k_u', 'k_cu', 's_du',
    'pi_g', 'pi_n',
    'alpha_g', 'alpha_n'
]

def pack_params(p):
    return np.array([p[name] for name in PARAM_NAMES], dtype=float)

def unpack_params(x):
    return {name: float(val) for name, val in zip(PARAM_NAMES, x)}

# =========================================
# Residuals for one dose
# =========================================
def residuals_one_dose(x, dose):
    p = unpack_params(x)

    # sanity checks
    if p['k_u'] <= 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if p['k_cu'] < 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if p['w_dc'] <= 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['s_du'] <= 1):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['pi_g'] <= 0.5):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['pi_n'] <= 0.8):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha_g'] <= 0.95):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha_n'] <= 0.95):
        return np.ones(4 * len(t_exp)) * 1e3

    sol_g = simulate_cplus(dose, 'gamma', t_exp, p)
    sol_n = simulate_cplus(dose, 'neutron', t_exp, p)

    if not sol_g.success or not sol_n.success:
        return np.ones(4 * len(t_exp)) * 1e3

    dg = DATA[dose]['gamma']
    dn = DATA[dose]['neutron']

    r = []
    r.extend((sol_g.y[5] - dg['gh2ax']) / (dg['gh2ax_err'] + 1e-12))
    r.extend(((sol_g.y[3] + sol_g.y[4]) - dg['patm']) / (dg['patm_err'] + 1e-12))
    r.extend((sol_n.y[5] - dn['gh2ax']) / (dn['gh2ax_err'] + 1e-12))
    r.extend(((sol_n.y[3] + sol_n.y[4]) - dn['patm']) / (dn['patm_err'] + 1e-12))
    return np.array(r)

# =========================================
# Initial guess and bounds
# =========================================
p0 = {
    'k_rs': 1.5,
    'V_rc': 4.4,
    'K_rc': 2.2,
    'V_af': 156.0,
    'K_m': 1.7,
    'k_df': 5.7,
    'k_as': 0.85,
    'k_ds': 0.26,
    'k_p': 16.6,
    'k_g': 8.5,
    'w_dc': 1.6,
    'k_u': 0.026,
    'k_cu': 0.10,
    's_du': 0.47,
    'pi_g': 0.02,
    'pi_n': 0.25,
    'alpha_g': 0.30,
    'alpha_n': 0.60,
}
x0 = pack_params(p0)

lb = np.array([
    1e-3, 1e-3, 1e-4,
    1e-2, 1e-4, 1e-3, 1e-4, 1e-4, 1e-3, 1e-3,
    0.1, 1e-4, 0.0, 0.0,
    0.0, 0.0,
    0.02, 0.02
], dtype=float)

ub = np.array([
    5.0, 10.0, 10.0,
    500.0, 10.0, 20.0, 5.0, 1.0, 50.0, 20.0,
    10.0, 5e-2, 5e-1, 1.0,
    0.5, 0.8,
    0.95, 0.95
], dtype=float)

print("len(x0) =", len(x0))
print("len(lb) =", len(lb))
print("len(ub) =", len(ub))

# =========================================
# Multi-start seeds for alpha_g / alpha_n / pi_n
# =========================================
starts = [
    {'alpha_g': 0.20, 'alpha_n': 0.45, 'pi_n': 0.20},
    {'alpha_g': 0.25, 'alpha_n': 0.55, 'pi_n': 0.25},
    {'alpha_g': 0.30, 'alpha_n': 0.60, 'pi_n': 0.25},
    {'alpha_g': 0.35, 'alpha_n': 0.70, 'pi_n': 0.35},
    {'alpha_g': 0.30, 'alpha_n': 0.75, 'pi_n': 0.45},
]
# dose-specific starting guesses for k_cu
kcu_start_by_dose = {
    0.25: 0.22,
    0.5: 0.20,
    1.0: 0.10,
    2.0: 0.03,
}

# =========================================
# Fit each dose separately
# =========================================
all_results = {}
summary_rows = []
init_rows = []
chi2_rows = []

for dose in DOSES:
    print(f"\n{'='*70}\nFITTING DOSE = {dose} Gy\n{'='*70}")

    best_res = None
    best_p = None
    best_chi2 = np.inf
    best_aic = np.inf
    best_seed = None

    ms_rows = []

    for seed in starts:
        p_start = p0.copy()
        p_start['alpha_g'] = seed['alpha_g']
        p_start['alpha_n'] = seed['alpha_n']
        p_start['pi_n'] = seed['pi_n']
        p_start['k_cu'] = kcu_start_by_dose[dose]

        x0_try = pack_params(p_start)

        res_try = least_squares(
            lambda x: residuals_one_dose(x, dose),
            x0_try,
            bounds=(lb, ub),
            method='trf',
            max_nfev=30000,
            ftol=1e-8,
            xtol=1e-8,
            gtol=1e-8,
            verbose=0
        )

        p_try = unpack_params(res_try.x)
        chi2_try = float(np.sum(res_try.fun**2))
        aic_try = 2 * len(res_try.x) + chi2_try

        ms_rows.append({
            'dose': dose,
            'alpha_g_start': seed['alpha_g'],
            'alpha_n_start': seed['alpha_n'],
            'pi_n_start': seed['pi_n'],
            'success': bool(res_try.success),
            'chi2': chi2_try,
            'AIC': aic_try,
            'alpha_g_fit': p_try['alpha_g'],
            'alpha_n_fit': p_try['alpha_n'],
            'pi_g_fit': p_try['pi_g'],
            'pi_n_fit': p_try['pi_n'],
            'k_cu_fit': p_try['k_cu'],
        })

        if chi2_try < best_chi2:
            best_res = res_try
            best_p = p_try
            best_chi2 = chi2_try
            best_aic = aic_try
            best_seed = seed

    df_ms = pd.DataFrame(ms_rows).sort_values("chi2").reset_index(drop=True)
    print("\n--- Multi-start summary ---")
    display(df_ms)

    print("\n--- Best fit ---")
    print("Best start =", best_seed)
    print("chi2 =", best_chi2)
    print("AIC  =", best_aic)
    for key in PARAM_NAMES:
        print(f"{key:>8s} = {best_p[key]:8.5f}")

    all_results[dose] = {
        'res': best_res,
        'params': best_p,
        'chi2': best_chi2,
        'AIC': best_aic,
        'multistart': df_ms,
    }

    # summary row
    row = {'dose': dose, 'chi2': best_chi2, 'AIC': best_aic}
    row.update(best_p)
    summary_rows.append(row)

    # initial fractions
    for rad in ['gamma', 'neutron']:
        alpha_fast = best_p['alpha_g'] if rad == 'gamma' else best_p['alpha_n']
        pi_used = best_p['pi_g'] if rad == 'gamma' else best_p['pi_n']

        slow_pool = 1.0 - alpha_fast
        Du0_fraction = pi_used * slow_pool
        Dc0_fraction = (1.0 - pi_used) * slow_pool

        init_rows.append({
            'dose': dose,
            'radiation': rad,
            'alpha_fast': alpha_fast,
            'slow_pool': slow_pool,
            'pi_used': pi_used,
            'Dc0_fraction': Dc0_fraction,
            'Du0_fraction': Du0_fraction
        })

    # residuals and chi2 contributions
    p = best_p
    sol_g = simulate_cplus(dose, 'gamma', t_exp, p)
    sol_n = simulate_cplus(dose, 'neutron', t_exp, p)

    dg = DATA[dose]['gamma']
    dn = DATA[dose]['neutron']

    z_gh_g = (sol_g.y[5] - dg['gh2ax']) / (dg['gh2ax_err'] + 1e-12)
    z_pa_g = ((sol_g.y[3] + sol_g.y[4]) - dg['patm']) / (dg['patm_err'] + 1e-12)
    z_gh_n = (sol_n.y[5] - dn['gh2ax']) / (dn['gh2ax_err'] + 1e-12)
    z_pa_n = ((sol_n.y[3] + sol_n.y[4]) - dn['patm']) / (dn['patm_err'] + 1e-12)

    print("\n--- Node-wise z-residuals ---")
    print(f"Dose {dose} Gy | gamma")
    print("  gH2AX:", np.round(z_gh_g, 2))
    print("  pATM :", np.round(z_pa_g, 2))
    print(f"Dose {dose} Gy | neutron")
    print("  gH2AX:", np.round(z_gh_n, 2))
    print("  pATM :", np.round(z_pa_n, 2))

    chi2_rows.extend([
        {"dose": dose, "radiation": "gamma",   "marker": "gH2AX", "chi2": float(np.sum(z_gh_g**2))},
        {"dose": dose, "radiation": "gamma",   "marker": "pATM",  "chi2": float(np.sum(z_pa_g**2))},
        {"dose": dose, "radiation": "neutron", "marker": "gH2AX", "chi2": float(np.sum(z_gh_n**2))},
        {"dose": dose, "radiation": "neutron", "marker": "pATM",  "chi2": float(np.sum(z_pa_n**2))},
    ])

    # plots
    t_dense = np.linspace(0.5, 24, 200)
    sol_g_dense = simulate_cplus(dose, 'gamma', t_dense, p)
    sol_n_dense = simulate_cplus(dose, 'neutron', t_dense, p)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

    # γH2AX
    axes[0].plot(t_dense, sol_g_dense.y[5], '-', lw=2, color='blue', label='Модель, γ')
    axes[0].errorbar(t_exp, dg['gh2ax'], yerr=dg['gh2ax_err'],
                     fmt='o', color='blue', capsize=4, label='Эксперимент, γ')

    axes[0].plot(t_dense, sol_n_dense.y[5], '--', lw=2, color='red', label='Модель, нейтроны')
    axes[0].errorbar(t_exp, dn['gh2ax'], yerr=dn['gh2ax_err'],
                     fmt='s', color='red', capsize=4, label='Эксперимент, n')

    axes[0].set_xlabel('Время (ч)')
    axes[0].set_ylabel(r'$\gamma$H2AX (отн.ед.)')
    axes[0].set_title(f'{dose} Гр')
    axes[0].legend()

    # pATM
    axes[1].plot(t_dense, sol_g_dense.y[3] + sol_g_dense.y[4], '-', lw=2, color='blue', label='Модель, γ')
    axes[1].errorbar(t_exp, dg['patm'], yerr=dg['patm_err'],
                     fmt='o', color='blue', capsize=4, label='Эксперимент, γ')

    axes[1].plot(t_dense, sol_n_dense.y[3] + sol_n_dense.y[4], '--', lw=2, color='red', label='Модель, нейтроны')
    axes[1].errorbar(t_exp, dn['patm'], yerr=dn['patm_err'],
                     fmt='s', color='red', capsize=4, label='Эксперимент, нейтроны')

    axes[1].set_xlabel('Время (ч)')
    axes[1].set_ylabel('pATM (отн.ед.)')
    axes[0].set_title(f'{dose} Гр')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig(os.path.join(SAVE_DIR, f'fit_{dose}Gy.png'), dpi=300)
    plt.show()

# =========================================
# Final tables
# =========================================
df_summary = pd.DataFrame(summary_rows)
print("\n=== Final parameter table ===")
display(df_summary)
df_summary.to_csv(os.path.join(SAVE_DIR, 'parameters_per_dose.csv'),
                  index=False, float_format='%.6f')

df_init = pd.DataFrame(init_rows)
print("\n=== Initial fractions table ===")
display(df_init)
df_init.to_csv(os.path.join(SAVE_DIR, 'initial_fractions_per_dose.csv'),
               index=False, float_format='%.6f')

df_chi2 = pd.DataFrame(chi2_rows)
print("\n=== Chi2 contributions by series ===")
display(df_chi2)
df_chi2.to_csv(os.path.join(SAVE_DIR, 'chi2_contributions_per_dose.csv'),
               index=False, float_format='%.6f')

# save multistart tables
for dose in DOSES:
    all_results[dose]['multistart'].to_csv(
        os.path.join(SAVE_DIR, f'multistart_{dose}Gy.csv'),
        index=False, float_format='%.6f'
    )

# save best params as json
json_ready = {}
for dose in DOSES:
    json_ready[str(dose)] = {
        'chi2': all_results[dose]['chi2'],
        'AIC': all_results[dose]['AIC'],
        'params': all_results[dose]['params']
    }

with open(os.path.join(SAVE_DIR, 'best_fits_per_dose.json'), 'w') as f:
    json.dump(json_ready, f, indent=2)
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]
PERT_FRACS = [0.05, 0.10]

param_to_lb = {p: lb[i] for i, p in enumerate(PARAM_NAMES)}
param_to_ub = {p: ub[i] for i, p in enumerate(PARAM_NAMES)}

def residuals_one_dose_dataset(x, dose, data_dose):
    p = unpack_params(x)

    if p['k_u'] <= 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if p['k_cu'] < 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if p['w_dc'] <= 0:
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['s_du'] <= 1):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['pi_g'] <= 0.5):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0 <= p['pi_n'] <= 0.8):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha_g'] <= 0.95):
        return np.ones(4 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha_n'] <= 0.95):
        return np.ones(4 * len(t_exp)) * 1e3

    sol_g = simulate_cplus(dose, 'gamma', t_exp, p)
    sol_n = simulate_cplus(dose, 'neutron', t_exp, p)

    if not sol_g.success or not sol_n.success:
        return np.ones(4 * len(t_exp)) * 1e3

    dg = data_dose['gamma']
    dn = data_dose['neutron']

    r = []
    r.extend((sol_g.y[5] - dg['gh2ax']) / (dg['gh2ax_err'] + 1e-12))
    r.extend(((sol_g.y[3] + sol_g.y[4]) - dg['patm']) / (dg['patm_err'] + 1e-12))
    r.extend((sol_n.y[5] - dn['gh2ax']) / (dn['gh2ax_err'] + 1e-12))
    r.extend(((sol_n.y[3] + sol_n.y[4]) - dn['patm']) / (dn['patm_err'] + 1e-12))
    return np.array(r)

sens_rows = []

for dose in DOSES_ANALYSIS:
    p_mle = all_results[dose]['params']
    x_mle = pack_params(p_mle)
    chi2_mle = all_results[dose]['chi2']

    for param in PARAM_NAMES:
        theta0 = p_mle[param]

        for frac in PERT_FRACS:
            for sign in [-1, 1]:
                label = f"{'-' if sign < 0 else '+'}{int(frac*100)}%"
                x_test = x_mle.copy()

                if abs(theta0) < 1e-12:
                    theta_new = theta0 + sign * frac
                else:
                    theta_new = theta0 * (1.0 + sign * frac)

                theta_new = np.clip(theta_new, param_to_lb[param], param_to_ub[param])

                idx = PARAM_NAMES.index(param)
                x_test[idx] = theta_new

                chi2_test = np.sum(residuals_one_dose_dataset(x_test, dose, DATA[dose])**2)
                dchi2 = chi2_test - chi2_mle
                sens_norm = abs(dchi2) / (chi2_mle * frac + 1e-12)

                sens_rows.append({
                    "dose": dose,
                    "param": param,
                    "theta_MLE": theta0,
                    "perturbation": label,
                    "theta_test": theta_new,
                    "chi2_MLE": chi2_mle,
                    "chi2_test": chi2_test,
                    "dchi2": dchi2,
                    "sens_norm": sens_norm,
                })

df_sens = pd.DataFrame(sens_rows)
print("=== Sensitivity table ===")
display(df_sens.head(20))

# сводка по максимальной чувствительности
df_sens_summary = (
    df_sens.groupby(["dose", "param"], as_index=False)
    .agg(
        theta_MLE=("theta_MLE", "first"),
        max_abs_dchi2=("dchi2", lambda s: np.max(np.abs(s))),
        max_sens_norm=("sens_norm", "max")
    )
    .sort_values(["dose", "max_sens_norm"], ascending=[True, False])
)

print("=== Max sensitivity by dose ===")
display(df_sens_summary)
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import least_squares
from IPython.display import display

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]
N_BOOT = 100
BOOT_SEED = 42

rng = np.random.default_rng(BOOT_SEED)

def make_boot_dataset(dose, rng):
    boot = {"gamma": {}, "neutron": {}}
    for rad in ["gamma", "neutron"]:
        for marker in ["gh2ax", "patm"]:
            mu = DATA[dose][rad][marker]
            sigma = DATA[dose][rad][f"{marker}_err"]
            y = rng.normal(mu, sigma)
            y = np.clip(y, 0.0, None)

            boot[rad][marker] = y
            boot[rad][f"{marker}_err"] = sigma.copy()
    return boot

boot_rows = []
boot_fail_rows = []

for dose in DOSES_ANALYSIS:
    print(f"\n=== Bootstrap for dose {dose} Gy ===")
    p_mle = all_results[dose]['params']
    x_start = pack_params(p_mle)

    for b in range(N_BOOT):
        boot_data = make_boot_dataset(dose, rng)

        try:
            res_boot = least_squares(
                lambda x: residuals_one_dose_dataset(x, dose, boot_data),
                x_start,
                bounds=(lb, ub),
                method='trf',
                max_nfev=20000,
                ftol=1e-8,
                xtol=1e-8,
                gtol=1e-8,
                verbose=0
            )

            if (not res_boot.success) or (not np.all(np.isfinite(res_boot.x))):
                boot_fail_rows.append({"dose": dose, "boot_id": b, "reason": "fit_failed"})
                continue

            p_boot = unpack_params(res_boot.x)
            chi2_boot = float(np.sum(res_boot.fun**2))

            row = {
                "dose": dose,
                "boot_id": b,
                "chi2": chi2_boot
            }
            row.update(p_boot)
            boot_rows.append(row)

        except Exception as e:
            boot_fail_rows.append({"dose": dose, "boot_id": b, "reason": str(e)})

    done = sum(r["dose"] == dose for r in boot_rows)
    failed = sum(r["dose"] == dose for r in boot_fail_rows)
    print(f"done={done}, failed={failed}")

boot_df = pd.DataFrame(boot_rows)
boot_fail_df = pd.DataFrame(boot_fail_rows)

print("\n=== Bootstrap samples ===")
display(boot_df.head())

print("\n=== Bootstrap failures ===")
display(boot_fail_df.head())

boot_df.to_csv("bootstrap_per_dose_params.csv", index=False)
boot_fail_df.to_csv("bootstrap_per_dose_failures.csv", index=False)
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

summary_rows = []

for param in KEY_PARAMS:
    means = []
    cis = []

    for dose in DOSES_ANALYSIS:
        vals = boot_df.loc[
            boot_df["dose"] == dose, param
        ].dropna().values

        if len(vals) == 0:
            continue

        mean = np.mean(vals)
        q025 = np.percentile(vals, 2.5)
        q975 = np.percentile(vals, 97.5)

        means.append(mean)
        cis.append((q025, q975))

    if len(means) < 2:
        continue

    rel_change = (max(means) - min(means)) / (abs(np.mean(means)) + 1e-12)

    overlap_count = 0
    for i in range(len(cis) - 1):
        lo1, hi1 = cis[i]
        lo2, hi2 = cis[i + 1]

        overlap = not (hi1 < lo2 or hi2 < lo1)
        overlap_count += int(overlap)

    summary_rows.append({
        "param": param,
        "mean_range": f"{min(means):.4g} → {max(means):.4g}",
        "relative_change": rel_change,
        "adjacent_CI_overlaps": overlap_count,
        "max_possible_overlap": len(cis) - 1
    })

df_text_joint = pd.DataFrame(summary_rows)
df_text_joint = df_text_joint.sort_values("relative_change", ascending=False).reset_index(drop=True)

print("=== Parameter trend summary across doses (joint gamma + neutron fit) ===")
display(df_text_joint)

df_text_joint.to_csv("parameter_trend_summary_joint_fit.csv", index=False)
```
