```python
# -*- coding: utf-8 -*-
"""
C+ model: separate fit for each dose
PROTONS 150 MeV ONLY
NO 0.1 Gy

For each dose:
- fit gh2ax and patm jointly
- one common parameter set within this dose
- one initial fast fraction alpha
- one ultraslow fraction pi

States:
    [D_s, D_c, D_u, pATM_f, pATM_s, gH2AX]
"""

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import least_squares
import pandas as pd
import os
import json
from IPython.display import display

# =========================================
# Settings
# =========================================
SAVE_DIR = "./Cplus_per_dose_protons_150MeV"
os.makedirs(SAVE_DIR, exist_ok=True)

plt.rcParams.update({
    'figure.figsize': (10, 5),
    'font.size': 12,
    'axes.labelsize': 13,
    'axes.grid': True,
    'grid.alpha': 0.3,
    'legend.frameon': False
})

# =========================================
# Experimental data: 150 MeV protons
# =========================================
t_exp = np.array([0.5, 2.0, 4.0, 6.0, 24.0], dtype=float)
DOSES = [0.25, 0.5, 1.0, 2.0]

DATA = {
    0.25: {
        "gh2ax":     np.array([8.81, 7.37, 5.82, 5.17, 3.65], dtype=float),
        "gh2ax_err": np.array([0.93, 0.56, 0.46, 0.40, 0.49], dtype=float),
        "patm":      np.array([3.99, 3.35, 2.36, 1.99, 1.30], dtype=float),
        "patm_err":  np.array([0.36, 0.31, 0.22, 0.19, 0.30], dtype=float),
    },

    0.5: {
        "gh2ax":     np.array([12.09, 9.51, 7.56, 6.73, 3.22], dtype=float),
        "gh2ax_err": np.array([0.76, 0.38, 0.67, 0.37, 0.28], dtype=float),
        "patm":      np.array([6.72, 4.72, 3.78, 3.32, 1.06], dtype=float),
        "patm_err":  np.array([0.41, 0.64, 0.245, 0.51, 0.35], dtype=float),
    },

    1.0: {
        "gh2ax":     np.array([19.92, 14.97, 10.76, 8.21, 3.24], dtype=float),
        "gh2ax_err": np.array([1.26, 0.66, 2.80, 0.71, 0.63], dtype=float),
        "patm":      np.array([10.81, 8.56, 5.36, 4.23, 1.82], dtype=float),
        "patm_err":  np.array([0.37, 0.52, 0.24, 0.63, 0.37], dtype=float),
    },

    2.0: {
        "gh2ax":     np.array([32.13, 24.45, 16.89, 12.65, 3.48], dtype=float),
        "gh2ax_err": np.array([2.15, 1.54, 2.47, 1.22, 0.81], dtype=float),
        "patm":      np.array([19.63, 14.32, 9.41, 7.05, 2.24], dtype=float),
        "patm_err":  np.array([1.31, 0.89, 1.03, 0.75, 0.29], dtype=float),
    },
}

# =========================================
# Model C+
# =========================================
def rhs_cplus(t, y, p):
    D_s, D_c, D_u, pATM_f, pATM_s, gH2AX = y

    dD_s = -p['k_rs'] * D_s
    dD_c = -(p['V_rc'] * D_c) / (p['K_rc'] + D_c + 1e-12) - p['k_cu'] * D_c
    dD_u = -p['k_u'] * D_u + p['k_cu'] * D_c

    w_du = p['s_du'] * p['w_dc']
    weighted = D_s + p['w_dc'] * D_c + w_du * D_u

    prod_f = (p['V_af'] * weighted) / (p['K_m'] + weighted + 1e-12)
    dpATM_f = prod_f - p['k_df'] * pATM_f
    dpATM_s = p['k_as'] * pATM_f - p['k_ds'] * pATM_s

    dgH2AX = p['k_p'] * (pATM_f + pATM_s) - p['k_g'] * gH2AX

    return [dD_s, dD_c, dD_u, dpATM_f, dpATM_s, dgH2AX]


def simulate_cplus(dose, t_eval, p):
    alpha = float(np.clip(p['alpha'], 0.02, 0.95))
    pi = float(np.clip(p['pi'], 0.0, 0.95))

    D_s0 = alpha * dose
    D_slow0 = (1.0 - alpha) * dose

    D_u0 = pi * D_slow0
    D_c0 = (1.0 - pi) * D_slow0

    y0 = [D_s0, D_c0, D_u0, 0.0, 0.0, 0.0]

    sol = solve_ivp(
        lambda t, y: rhs_cplus(t, y, p),
        (0.0, float(np.max(t_eval))),
        y0,
        t_eval=t_eval,
        method='LSODA',
        rtol=1e-6,
        atol=1e-8
    )
    return sol

# =========================================
# Parameterization for one dose
# =========================================
PARAM_NAMES = [
    'k_rs', 'V_rc', 'K_rc',
    'V_af', 'K_m', 'k_df', 'k_as', 'k_ds', 'k_p', 'k_g',
    'w_dc', 'k_u', 'k_cu', 's_du',
    'pi', 'alpha'
]

def pack_params(p):
    return np.array([p[name] for name in PARAM_NAMES], dtype=float)

def unpack_params(x):
    return {name: float(val) for name, val in zip(PARAM_NAMES, x)}

# =========================================
# Residuals for one dose
# =========================================
def residuals_one_dose(x, dose):
    p = unpack_params(x)

    # sanity checks
    if p['k_u'] <= 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if p['k_cu'] < 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if p['w_dc'] <= 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0 <= p['s_du'] <= 1):
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0 <= p['pi'] <= 0.8):
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha'] <= 0.95):
        return np.ones(2 * len(t_exp)) * 1e3

    sol = simulate_cplus(dose, t_exp, p)
    if not sol.success:
        return np.ones(2 * len(t_exp)) * 1e3

    d = DATA[dose]

    r = []
    r.extend((sol.y[5] - d['gh2ax']) / (d['gh2ax_err'] + 1e-12))
    r.extend(((sol.y[3] + sol.y[4]) - d['patm']) / (d['patm_err'] + 1e-12))
    return np.array(r)

# =========================================
# Initial guess and bounds
# =========================================
p0 = {
    'k_rs': 1.5,
    'V_rc': 4.4,
    'K_rc': 2.2,
    'V_af': 156.0,
    'K_m': 1.7,
    'k_df': 5.7,
    'k_as': 0.85,
    'k_ds': 0.26,
    'k_p': 16.6,
    'k_g': 8.5,
    'w_dc': 1.6,
    'k_u': 0.026,
    'k_cu': 0.10,
    's_du': 0.47,
    'pi': 0.20,
    'alpha': 0.45,
}
x0 = pack_params(p0)

lb = np.array([
    1e-3, 1e-3, 1e-4,
    1e-2, 1e-4, 1e-3, 1e-4, 1e-4, 1e-3, 1e-3,
    0.1, 1e-4, 0.0, 0.0,
    0.0, 0.02
], dtype=float)

ub = np.array([
    5.0, 10.0, 10.0,
    500.0, 10.0, 20.0, 5.0, 1.0, 50.0, 20.0,
    10.0, 5e-2, 5e-1, 1.0,
    0.8, 0.95
], dtype=float)

print("len(x0) =", len(x0))
print("len(lb) =", len(lb))
print("len(ub) =", len(ub))

# =========================================
# Multi-start seeds for alpha / pi
# =========================================
starts = [
    {'alpha': 0.30, 'pi': 0.10},
    {'alpha': 0.40, 'pi': 0.15},
    {'alpha': 0.45, 'pi': 0.20},
    {'alpha': 0.55, 'pi': 0.25},
    {'alpha': 0.65, 'pi': 0.35},
]

# dose-specific starting guesses for k_cu
kcu_start_by_dose = {
    0.25: 0.20,
    0.5: 0.15,
    1.0: 0.10,
    2.0: 0.05,
}

# =========================================
# Fit each dose separately
# =========================================
all_results = {}
summary_rows = []
init_rows = []
chi2_rows = []

for dose in DOSES:
    print(f"\n{'='*70}\nFITTING DOSE = {dose} Gy\n{'='*70}")

    best_res = None
    best_p = None
    best_chi2 = np.inf
    best_aic = np.inf
    best_seed = None

    ms_rows = []

    for seed in starts:
        p_start = p0.copy()
        p_start['alpha'] = seed['alpha']
        p_start['pi'] = seed['pi']
        p_start['k_cu'] = kcu_start_by_dose[dose]

        x0_try = pack_params(p_start)

        res_try = least_squares(
            lambda x: residuals_one_dose(x, dose),
            x0_try,
            bounds=(lb, ub),
            method='trf',
            max_nfev=30000,
            ftol=1e-8,
            xtol=1e-8,
            gtol=1e-8,
            verbose=0
        )

        p_try = unpack_params(res_try.x)
        chi2_try = float(np.sum(res_try.fun**2))
        aic_try = 2 * len(res_try.x) + chi2_try

        ms_rows.append({
            'dose': dose,
            'alpha_start': seed['alpha'],
            'pi_start': seed['pi'],
            'success': bool(res_try.success),
            'chi2': chi2_try,
            'AIC': aic_try,
            'alpha_fit': p_try['alpha'],
            'pi_fit': p_try['pi'],
            'k_cu_fit': p_try['k_cu'],
        })

        if chi2_try < best_chi2:
            best_res = res_try
            best_p = p_try
            best_chi2 = chi2_try
            best_aic = aic_try
            best_seed = seed

    df_ms = pd.DataFrame(ms_rows).sort_values("chi2").reset_index(drop=True)
    print("\n--- Multi-start summary ---")
    display(df_ms)

    print("\n--- Best fit ---")
    print("Best start =", best_seed)
    print("chi2 =", best_chi2)
    print("AIC  =", best_aic)
    for key in PARAM_NAMES:
        print(f"{key:>8s} = {best_p[key]:8.5f}")

    all_results[dose] = {
        'res': best_res,
        'params': best_p,
        'chi2': best_chi2,
        'AIC': best_aic,
        'multistart': df_ms,
    }

    # summary row
    row = {'dose': dose, 'chi2': best_chi2, 'AIC': best_aic}
    row.update(best_p)
    summary_rows.append(row)

    # initial fractions
    alpha_fast = best_p['alpha']
    slow_pool = 1.0 - alpha_fast
    pi_used = best_p['pi']

    Du0_fraction = pi_used * slow_pool
    Dc0_fraction = (1.0 - pi_used) * slow_pool

    init_rows.append({
        'dose': dose,
        'alpha_fast': alpha_fast,
        'slow_pool': slow_pool,
        'pi_used': pi_used,
        'Dc0_fraction': Dc0_fraction,
        'Du0_fraction': Du0_fraction
    })

    # residuals and chi2 contributions
    p = best_p
    sol = simulate_cplus(dose, t_exp, p)
    d = DATA[dose]

    z_gh = (sol.y[5] - d['gh2ax']) / (d['gh2ax_err'] + 1e-12)
    z_pa = ((sol.y[3] + sol.y[4]) - d['patm']) / (d['patm_err'] + 1e-12)

    print("\n--- Node-wise z-residuals ---")
    print(f"Dose {dose} Gy | proton")
    print("  gH2AX:", np.round(z_gh, 2))
    print("  pATM :", np.round(z_pa, 2))

    chi2_rows.extend([
        {"dose": dose, "marker": "gH2AX", "chi2": float(np.sum(z_gh**2))},
        {"dose": dose, "marker": "pATM",  "chi2": float(np.sum(z_pa**2))},
    ])

    # plots
    t_dense = np.linspace(0.5, 24, 200)
    sol_dense = simulate_cplus(dose, t_dense, p)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

    # γH2AX
    axes[0].plot(t_dense, sol_dense.y[5], '-', lw=2, color='blue', label='Модель')
    axes[0].errorbar(
        t_exp, d['gh2ax'], yerr=d['gh2ax_err'],
        fmt='o', color='blue', capsize=4, label='Эксперимент'
    )
    axes[0].set_xlabel('Время (ч)')
    axes[0].set_ylabel(r'$\gamma$H2AX (отн.ед.)')
    axes[0].set_title(f'Протоны, {dose} Гр')
    axes[0].legend()

    # pATM
    axes[1].plot(t_dense, sol_dense.y[3] + sol_dense.y[4], '-', lw=2, color='blue', label='Модель')
    axes[1].errorbar(
        t_exp, d['patm'], yerr=d['patm_err'],
        fmt='s', color='blue', capsize=4, label='Эксперимент'
    )
    axes[1].set_xlabel('Время (ч)')
    axes[1].set_ylabel('pATM (отн.ед.)')
    axes[1].set_title(f'Протоны, {dose} Гр')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig(os.path.join(SAVE_DIR, f'fit_{dose}Gy.png'), dpi=300)
    plt.show()

# =========================================
# Final tables
# =========================================
df_summary = pd.DataFrame(summary_rows)
print("\n=== Final parameter table ===")
display(df_summary)
df_summary.to_csv(
    os.path.join(SAVE_DIR, 'parameters_per_dose.csv'),
    index=False, float_format='%.6f'
)

df_init = pd.DataFrame(init_rows)
print("\n=== Initial fractions table ===")
display(df_init)
df_init.to_csv(
    os.path.join(SAVE_DIR, 'initial_fractions_per_dose.csv'),
    index=False, float_format='%.6f'
)

df_chi2 = pd.DataFrame(chi2_rows)
print("\n=== Chi2 contributions by series ===")
display(df_chi2)
df_chi2.to_csv(
    os.path.join(SAVE_DIR, 'chi2_contributions_per_dose.csv'),
    index=False, float_format='%.6f'
)

# save multistart tables
for dose in DOSES:
    all_results[dose]['multistart'].to_csv(
        os.path.join(SAVE_DIR, f'multistart_{dose}Gy.csv'),
        index=False, float_format='%.6f'
    )

# save best params as json
json_ready = {}
for dose in DOSES:
    json_ready[str(dose)] = {
        'chi2': all_results[dose]['chi2'],
        'AIC': all_results[dose]['AIC'],
        'params': all_results[dose]['params']
    }

with open(os.path.join(SAVE_DIR, 'best_fits_per_dose.json'), 'w') as f:
    json.dump(json_ready, f, indent=2)
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]
PERT_FRACS = [0.05, 0.10]

param_to_lb = {p: lb[i] for i, p in enumerate(PARAM_NAMES)}
param_to_ub = {p: ub[i] for i, p in enumerate(PARAM_NAMES)}

def residuals_one_dose_dataset(x, dose, data_dose):
    p = unpack_params(x)

    if p['k_u'] <= 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if p['k_cu'] < 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if p['w_dc'] <= 0:
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0 <= p['s_du'] <= 1):
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0 <= p['pi'] <= 0.8):
        return np.ones(2 * len(t_exp)) * 1e3
    if not (0.02 <= p['alpha'] <= 0.95):
        return np.ones(2 * len(t_exp)) * 1e3

    sol = simulate_cplus(dose, t_exp, p)

    if not sol.success:
        return np.ones(2 * len(t_exp)) * 1e3

    d = data_dose

    r = []
    r.extend((sol.y[5] - d['gh2ax']) / (d['gh2ax_err'] + 1e-12))
    r.extend(((sol.y[3] + sol.y[4]) - d['patm']) / (d['patm_err'] + 1e-12))
    return np.array(r)

sens_rows = []

for dose in DOSES_ANALYSIS:
    p_mle = all_results[dose]['params']
    x_mle = pack_params(p_mle)
    chi2_mle = all_results[dose]['chi2']

    for param in PARAM_NAMES:
        theta0 = p_mle[param]

        for frac in PERT_FRACS:
            for sign in [-1, 1]:
                label = f"{'-' if sign < 0 else '+'}{int(frac * 100)}%"
                x_test = x_mle.copy()

                if abs(theta0) < 1e-12:
                    theta_new = theta0 + sign * frac
                else:
                    theta_new = theta0 * (1.0 + sign * frac)

                theta_new = np.clip(theta_new, param_to_lb[param], param_to_ub[param])

                idx = PARAM_NAMES.index(param)
                x_test[idx] = theta_new

                chi2_test = np.sum(residuals_one_dose_dataset(x_test, dose, DATA[dose]) ** 2)
                dchi2 = chi2_test - chi2_mle
                sens_norm = abs(dchi2) / (chi2_mle * frac + 1e-12)

                sens_rows.append({
                    "dose": dose,
                    "param": param,
                    "theta_MLE": theta0,
                    "perturbation": label,
                    "theta_test": theta_new,
                    "chi2_MLE": chi2_mle,
                    "chi2_test": chi2_test,
                    "dchi2": dchi2,
                    "sens_norm": sens_norm,
                })

df_sens = pd.DataFrame(sens_rows)
print("=== Sensitivity table ===")
display(df_sens)

# сводка по максимальной чувствительности
df_sens_summary = (
    df_sens.groupby(["dose", "param"], as_index=False)
    .agg(
        theta_MLE=("theta_MLE", "first"),
        max_abs_dchi2=("dchi2", lambda s: np.max(np.abs(s))),
        max_sens_norm=("sens_norm", "max")
    )
    .sort_values(["dose", "max_sens_norm"], ascending=[True, False])
)

print("=== Max sensitivity by dose ===")
display(df_sens_summary)
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import least_squares
from IPython.display import display

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]
N_BOOT = 100
BOOT_SEED = 42

rng = np.random.default_rng(BOOT_SEED)

def make_boot_dataset(dose, rng):
    boot = {}

    for marker in ["gh2ax", "patm"]:
        mu = DATA[dose][marker]
        sigma = DATA[dose][f"{marker}_err"]

        y = rng.normal(mu, sigma)
        y = np.clip(y, 0.0, None)

        boot[marker] = y
        boot[f"{marker}_err"] = sigma.copy()

    return boot

boot_rows = []
boot_fail_rows = []

for dose in DOSES_ANALYSIS:
    print(f"\n=== Bootstrap for dose {dose} Gy ===")
    p_mle = all_results[dose]['params']
    x_start = pack_params(p_mle)

    for b in range(N_BOOT):
        boot_data = make_boot_dataset(dose, rng)

        try:
            res_boot = least_squares(
                lambda x: residuals_one_dose_dataset(x, dose, boot_data),
                x_start,
                bounds=(lb, ub),
                method='trf',
                max_nfev=20000,
                ftol=1e-8,
                xtol=1e-8,
                gtol=1e-8,
                verbose=0
            )

            if (not res_boot.success) or (not np.all(np.isfinite(res_boot.x))):
                boot_fail_rows.append({
                    "dose": dose,
                    "boot_id": b,
                    "reason": "fit_failed"
                })
                continue

            p_boot = unpack_params(res_boot.x)
            chi2_boot = float(np.sum(res_boot.fun ** 2))

            row = {
                "dose": dose,
                "boot_id": b,
                "chi2": chi2_boot
            }
            row.update(p_boot)
            boot_rows.append(row)

        except Exception as e:
            boot_fail_rows.append({
                "dose": dose,
                "boot_id": b,
                "reason": str(e)
            })

    done = sum(r["dose"] == dose for r in boot_rows)
    failed = sum(r["dose"] == dose for r in boot_fail_rows)
    print(f"done={done}, failed={failed}")

boot_df = pd.DataFrame(boot_rows)
boot_fail_df = pd.DataFrame(boot_fail_rows)

print("\n=== Bootstrap samples ===")
display(boot_df.head())

print("\n=== Bootstrap failures ===")
display(boot_fail_df.head())

boot_df.to_csv("bootstrap_per_dose_params.csv", index=False)
boot_fail_df.to_csv("bootstrap_per_dose_failures.csv", index=False)
```

```python
import pandas as pd
from IPython.display import display

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]

# обновлено под proton-version модели
KEY_PARAMS = [
    "k_rs", "V_rc", "K_rc",
    "k_u", "k_cu", "s_du",
    "pi", "alpha",
    "w_dc", "k_as", "k_ds",
    "k_p", "k_g"
]

overlap_rows = []

for param in KEY_PARAMS:
    sub = (
        df_boot_summary[df_boot_summary["param"] == param]
        .sort_values("dose")
        .reset_index(drop=True)
    )

    if len(sub) < 2:
        continue

    for i in range(len(sub) - 1):
        d1 = sub.loc[i, "dose"]
        d2 = sub.loc[i + 1, "dose"]

        lo1, hi1 = sub.loc[i, "q2.5"], sub.loc[i, "q97.5"]
        lo2, hi2 = sub.loc[i + 1, "q2.5"], sub.loc[i + 1, "q97.5"]

        if pd.isna(lo1) or pd.isna(hi1) or pd.isna(lo2) or pd.isna(hi2):
            overlap = np.nan
        else:
            overlap = not (hi1 < lo2 or hi2 < lo1)

        overlap_rows.append({
            "param": param,
            "dose1": d1,
            "dose2": d2,
            "CI1": f"[{lo1:.4g}, {hi1:.4g}]" if pd.notna(lo1) and pd.notna(hi1) else "NA",
            "CI2": f"[{lo2:.4g}, {hi2:.4g}]" if pd.notna(lo2) and pd.notna(hi2) else "NA",
            "overlap": overlap
        })

df_overlap = pd.DataFrame(overlap_rows)

print("=== CI overlap table ===")
display(df_overlap)

df_overlap.to_csv("bootstrap_CI_overlap.csv", index=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt

DOSES_ANALYSIS = [0.25, 0.5, 1.0, 2.0]

# обновлено под proton-version модели
KEY_PARAMS = [
    "k_rs", "V_rc", "K_rc",
    "k_u", "k_cu", "s_du",
    "pi", "alpha",
    "w_dc", "k_as", "k_ds",
    "k_p", "k_g"
]

ncols = 3
nrows = int(np.ceil(len(KEY_PARAMS) / ncols))
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4.2 * nrows))
axes = np.atleast_1d(axes).ravel()

for i, param in enumerate(KEY_PARAMS):
    ax = axes[i]

    all_vals = []
    for dose in DOSES_ANALYSIS:
        if param not in boot_df.columns:
            continue
        vals = boot_df.loc[boot_df["dose"] == dose, param].dropna().values
        if len(vals) > 0:
            all_vals.append(vals)

    if len(all_vals) == 0:
        ax.set_title(param)
        ax.text(0.5, 0.5, "No data", ha="center", va="center")
        ax.axis("off")
        continue

    # общие bins
    vmin = min(np.min(v) for v in all_vals)
    vmax = max(np.max(v) for v in all_vals)

    if np.isclose(vmin, vmax):
        eps = 1e-6 if abs(vmin) < 1 else 0.01 * abs(vmin)
        bins = np.linspace(vmin - eps, vmax + eps, 25)
    else:
        bins = np.linspace(vmin, vmax, 25)

    for dose in DOSES_ANALYSIS:
        vals = boot_df.loc[boot_df["dose"] == dose, param].dropna().values
        if len(vals) == 0:
            continue
        ax.hist(vals, bins=bins, alpha=0.35, density=True, label=f"{dose} Gy")

    if param in ["pi", "alpha", "k_u", "k_cu", "s_du"]:
        ax.axvline(0.0, linestyle='-.', linewidth=1)

    ax.set_title(param)
    ax.set_xlabel(param)
    ax.set_ylabel("Density")
    ax.legend(fontsize=8)

for j in range(i + 1, len(axes)):
    axes[j].axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

summary_rows = []

for param in KEY_PARAMS:
    means = []
    cis = []

    for dose in DOSES_ANALYSIS:
        if param not in boot_df.columns:
            continue

        vals = boot_df.loc[
            boot_df["dose"] == dose, param
        ].dropna().values

        if len(vals) == 0:
            continue

        mean = np.mean(vals)
        q025 = np.percentile(vals, 2.5)
        q975 = np.percentile(vals, 97.5)

        means.append(mean)
        cis.append((q025, q975))

    if len(means) < 2:
        continue

    rel_change = (max(means) - min(means)) / (abs(np.mean(means)) + 1e-12)

    overlap_count = 0
    for i in range(len(cis) - 1):
        lo1, hi1 = cis[i]
        lo2, hi2 = cis[i + 1]

        overlap = not (hi1 < lo2 or hi2 < lo1)
        overlap_count += int(overlap)

    summary_rows.append({
        "param": param,
        "mean_range": f"{min(means):.4g} → {max(means):.4g}",
        "relative_change": rel_change,
        "adjacent_CI_overlaps": overlap_count,
        "max_possible_overlap": len(cis) - 1
    })

df_text = pd.DataFrame(summary_rows)
df_text = df_text.sort_values("relative_change", ascending=False)

print("=== Parameter trend summary across doses ===")
display(df_text)
```
